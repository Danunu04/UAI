**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file
states = set()
with open(datafile, 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        state = row["State"].strip()
        states.add(state)

print("Different states:", len(states))
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
from collections import defaultdict

state_count = defaultdict(int)

with open(datafile, 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        state = row["State"].strip()
        state_count[state] += 1

most_state = max(state_count, key=state_count.get)

print("State with most customers:", most_state)
print("Count:", state_count[most_state])
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file
least_state = min(state_count, key=state_count.get)

print("State with least customers:", least_state)
print("Count:", state_count[least_state])
```

**Question 5:** What is the most common last name?

```python
last_names = defaultdict(int)

with open(datafile, 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        last = row["LastName"].strip()
        last_names[last] += 1

most_common_last = max(last_names, key=last_names.get)

print("Most common last name:", most_common_last)
print("Count:", last_names[most_common_last])
```

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

```python
# Exercise 2: Processing Orders data

import csv
from collections import defaultdict
from datetime import datetime

orders_file = "./sample_data/orders.csv"

orders = defaultdict(list)

def parse_date(date_str):
    date_str = date_str.strip()
    if not date_str:
        return None

    formats = [
        "%Y-%m-%d %H:%M:%S.%f",
        "%Y-%m-%d %H:%M:%S"
    ]

    for fmt in formats:
        try:
            return datetime.strptime(date_str, fmt)
        except ValueError:
            pass

    return None

# Read and clean data
with open(orders_file, 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        try:
            customer_id = row["CustomerID"].strip()
            order_id = row["OrderID"].strip()
            date_str = row["Date"].strip()
            price_str = row["Price"].strip()

            if not customer_id or not order_id or not date_str:
                continue

            parsed_date = parse_date(date_str)
            if parsed_date is None:
                continue

            try:
                price = float(price_str)
            except ValueError:
                price = 0.0

            cleaned_row = {
                "CustomerID": customer_id,
                "OrderID": order_id,
                "Date": parsed_date,
                "Price": price
            }

            orders[order_id].append(cleaned_row)

        except:
            continue

# Question 1: How many unique orders are in the orders.csv file?
print("Unique orders:", len(orders))

# Question 2: What is the average number of items per order (rounded to two decimal places)?
total_items = sum(len(items) for items in orders.values())
avg_items = total_items / len(orders)
print("Average items per order:", round(avg_items, 2))

# Question 3: What is the highest number of items per order?
max_items = max(len(items) for items in orders.values())
print("Highest number of items in an order:", max_items)

# Question 4: What is the number of orders placed in October 2021?
oct_orders = 0

for order_id, items in orders.items():
    order_date = items[0]["Date"]
    if order_date.year == 2021 and order_date.month == 10:
        oct_orders += 1

print("Orders placed in October 2021:", oct_orders)

# Question 5: Which customer spent the most amount of money in 2021?
customer_spending = defaultdict(float)

for order_id, items in orders.items():
    order_date = items[0]["Date"]

    if order_date.year == 2021:
        customer_id = items[0]["CustomerID"]
        for item in items:
            customer_spending[customer_id] += item["Price"]

top_customer = max(customer_spending, key=customer_spending.get)
print("Customer who spent the most in 2021:", top_customer)
print("Amount spent:", round(customer_spending[top_customer], 2))

# Question 6: Historically, what is the best month for sales?
monthly_sales = defaultdict(float)

for order_id, items in orders.items():
    order_date = items[0]["Date"]
    month_key = f"{order_date.year}-{order_date.month:02d}"

    for item in items:
        monthly_sales[month_key] += item["Price"]

best_month = max(monthly_sales, key=monthly_sales.get)
print("Best month for sales:", best_month)
print("Total sales:", round(monthly_sales[best_month], 2))

print("Finished!")
```
